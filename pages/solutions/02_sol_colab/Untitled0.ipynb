## Sumowanie ciągu geometrycznego

Rozważmy ciąg geometryczny o pierwszym wyrazie $a_1$ oraz ilorazie $q$.

Suma pierwszych $n$ wyrazów ciągu geometrycznego dla $q \neq 1$ jest dana wzorem:

$$
S_n = a_1 \frac{1-q^n}{1-q}
$$

Przykład:

Dla:

- $a_1 = 2$
- $q = 3$
- $n = 4$

otrzymujemy:

$$
S_4 = 2 \cdot \frac{1-3^4}{1-3}
= 2 \cdot \frac{1-81}{-2}
= 80
$$

### Komórka kodu: wykres $y=k^2$ dla $k=1,\dots,20$

````python
import matplotlib.pyplot as plt

k = list(range(1, 21))
y = [i**2 for i in k]

plt.figure(figsize=(8,5))
plt.plot(k, y, marker="o")
plt.title("y = k²")
plt.xlabel("k")
plt.ylabel("y")
plt.grid(True)

plt.savefig("wykres.png", dpi=300, bbox_inches="tight")
plt.show()

print("Zapisano wykres.png")
````

### Komórka kodu: animacja wahadła MP4

````python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.animation as animation
from IPython.display import Video

L = 1.0
theta0 = np.pi / 4

fig, ax = plt.subplots(figsize=(5,5))
ax.set_xlim(-1.2, 1.2)
ax.set_ylim(-1.2, 0.2)
ax.set_aspect('equal')
ax.grid()

line, = ax.plot([], [], 'o-', lw=3)

def update(frame):
    theta = theta0 * np.cos(frame * 0.05)

    x = L * np.sin(theta)
    y = -L * np.cos(theta)

    line.set_data([0, x], [0, y])
    return line,

ani = animation.FuncAnimation(
    fig,
    update,
    frames=200,
    interval=30,
    blit=True
)

ani.save("video.mp4", writer="ffmpeg", fps=30)

plt.close()

Video("video.mp4", embed=True)
````

### Komórka kodu: zmodyfikowana krzywa parametryczna

Poniższa wersja generuje przestrzenną spiralę.

````python
import numpy as np
import matplotlib.pyplot as plt

t = np.linspace(0, 8*np.pi, 1000)

x = np.cos(t)
y = np.sin(t)
z = t / (2*np.pi)

fig = plt.figure(figsize=(8,8))
ax = fig.add_subplot(111, projection='3d')

ax.plot(
    x,
    y,
    z,
    color="crimson",
    linewidth=2,
    label="Helix"
)

ax.set_title("3D Helix Curve")
ax.set_xlabel("x(t)")
ax.set_ylabel("y(t")
ax.set_zlabel("z(t)")
ax.legend()

plt.savefig(
    "parametric_curve.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("Zapisano parametric_curve.png")
````
